```python
# ============================================================
# FIRMABLE - PROJECT SETUP
# ============================================================

from pyspark.sql import functions as F
from pyspark.sql.window import Window

# Project configuration
PROJECT_NAME = "Firmable"
PROJECT_VERSION = "2.0"

CATALOG = "workspace"

BRONZE_SCHEMA = "firmable_v2_bronze"
SILVER_SCHEMA = "firmable_v2_silver"
GOLD_SCHEMA = "firmable_v2_gold"
EVALUATION_SCHEMA = "firmable_v2_evaluation"

# Create schemas
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{BRONZE_SCHEMA}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{GOLD_SCHEMA}")
spark.sql(f"CREATE SCHEMA IF NOT EXISTS {CATALOG}.{EVALUATION_SCHEMA}")

# Display configuration
print("PROJECT SETUP COMPLETED")
print("=======================")
print(f"Project: {PROJECT_NAME}")
print(f"Version: {PROJECT_VERSION}")
print(f"Catalog: {CATALOG}")

print("\nMEDALLION SCHEMAS")
print("----------------")

for schema in [
    BRONZE_SCHEMA,
    SILVER_SCHEMA,
    GOLD_SCHEMA,
    EVALUATION_SCHEMA
]:
    print(f"{CATALOG}.{schema}")
```
