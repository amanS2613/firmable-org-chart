```python

# ============================================================
# NOTEBOOK 01: HIERARCHY GENERATION
# STEP 1: PROJECT AND HIERARCHY CONFIGURATION
# ============================================================

import random
import pandas as pd
import numpy as np

from datetime import datetime, timedelta

# ------------------------------------------------------------
# PROJECT CONFIGURATION
# ------------------------------------------------------------

NUM_COMPANIES = 500

MIN_EMPLOYEES_PER_COMPANY = 100
MAX_EMPLOYEES_PER_COMPANY = 250

RANDOM_SEED = 42

random.seed(RANDOM_SEED)
np.random.seed(RANDOM_SEED)


# ------------------------------------------------------------
# ORGANIZATIONAL SENIORITY
# Lower number = higher position
# CEO is the absolute top position
# ------------------------------------------------------------

SENIORITY_LEVELS = {
    "CEO": 0,
    "Executive": 1,
    "Vice President": 2,
    "Director": 3,
    "Senior Manager": 4,
    "Manager": 5,
    "Team Lead": 6,
    "Senior Individual Contributor": 7,
    "Individual Contributor": 8,
    "Junior": 9,
    "Intern": 10
}


# ------------------------------------------------------------
# JOB TITLES BY SENIORITY CATEGORY
# ------------------------------------------------------------

JOB_TITLES = {

    "CEO": [
        "Chief Executive Officer"
    ],

    "Executive": [
        "Chief Technology Officer",
        "Chief Financial Officer",
        "Chief Operating Officer",
        "Chief Data Officer",
        "Chief Marketing Officer",
        "Chief Human Resources Officer"
    ],

    "Vice President": [
        "Vice President of Engineering",
        "Vice President of Data",
        "Vice President of Sales",
        "Vice President of Marketing",
        "Vice President of Operations",
        "Vice President of Finance",
        "Vice President of Human Resources"
    ],

    "Director": [
        "Director of Engineering",
        "Director of Data",
        "Director of Sales",
        "Director of Marketing",
        "Director of Operations",
        "Director of Finance",
        "Director of Human Resources",
        "Director of Product"
    ],

    "Senior Manager": [
        "Senior Engineering Manager",
        "Senior Data Manager",
        "Senior Product Manager",
        "Senior Operations Manager",
        "Senior Finance Manager",
        "Senior Sales Manager"
    ],

    "Manager": [
        "Engineering Manager",
        "Data Engineering Manager",
        "Analytics Manager",
        "Product Manager",
        "Sales Manager",
        "Marketing Manager",
        "Finance Manager",
        "Operations Manager",
        "HR Manager"
    ],

    "Team Lead": [
        "Engineering Team Lead",
        "Data Team Lead",
        "Technical Lead",
        "Analytics Team Lead",
        "Product Team Lead"
    ],

    "Senior Individual Contributor": [
        "Senior Software Engineer",
        "Senior Data Engineer",
        "Senior Data Analyst",
        "Staff Software Engineer",
        "Principal Data Engineer",
        "Senior Business Analyst"
    ],

    "Individual Contributor": [
        "Software Engineer",
        "Data Engineer",
        "Data Analyst",
        "Business Analyst",
        "Product Analyst",
        "Marketing Analyst",
        "Financial Analyst"
    ],

    "Junior": [
        "Junior Software Engineer",
        "Junior Data Engineer",
        "Associate Analyst",
        "Associate Engineer",
        "Junior Business Analyst"
    ],

    "Intern": [
        "Software Engineering Intern",
        "Data Science Intern",
        "Business Analyst Intern",
        "Finance Intern",
        "Marketing Intern"
    ]
}


# ------------------------------------------------------------
# EMPLOYEES ELIGIBLE TO BECOME MANAGERS
# ------------------------------------------------------------

MANAGER_CATEGORIES = [
    "CEO",
    "Executive",
    "Vice President",
    "Director",
    "Senior Manager",
    "Manager",
    "Team Lead"
]


# ------------------------------------------------------------
# WORKFORCE DISTRIBUTION
# CEO is generated separately.
# These weights apply to all remaining employees.
# ------------------------------------------------------------

WORKFORCE_DISTRIBUTION = {
    "Executive": 0.015,
    "Vice President": 0.025,
    "Director": 0.045,
    "Senior Manager": 0.065,
    "Manager": 0.10,
    "Team Lead": 0.12,
    "Senior Individual Contributor": 0.14,
    "Individual Contributor": 0.25,
    "Junior": 0.14,
    "Intern": 0.10
}


# ------------------------------------------------------------
# ORGANIZATIONAL STRUCTURE TYPES
# Each company will receive one primary structure type.
# ------------------------------------------------------------

ORGANIZATIONAL_STRUCTURES = [
    "Functional",
    "Hierarchical",
    "Flat",
    "Divisional",
    "Matrix-Inspired"
]

STRUCTURE_WEIGHTS = [0.25, 0.25, 0.15, 0.20, 0.15]


print("Hierarchy configuration loaded successfully.")
print("Companies planned:", NUM_COMPANIES)
print("Seniority categories:", len(SENIORITY_LEVELS))
print("Job title categories:", len(JOB_TITLES))
print("Manager-eligible categories:", len(MANAGER_CATEGORIES))
print("Organizational structure types:", len(ORGANIZATIONAL_STRUCTURES))
```

```python

# ============================================================
# STEP 2: COMPANY AND ORGANIZATIONAL STRUCTURE CONFIGURATION
# ============================================================

# ------------------------------------------------------------
# COMPANY SIZE DISTRIBUTION
# ------------------------------------------------------------

COMPANY_SIZE_PROBABILITIES = {
    "Small": 0.40,
    "Medium": 0.35,
    "Large": 0.25
}

COMPANY_SIZE_PROFILES = {
    "Small": {
        "min_employees": 100,
        "max_employees": 124
    },
    "Medium": {
        "min_employees": 125,
        "max_employees": 174
    },
    "Large": {
        "min_employees": 175,
        "max_employees": 250
    }
}


# ------------------------------------------------------------
# INDUSTRIES
# ------------------------------------------------------------

INDUSTRIES = [
    "Technology",
    "Financial Services",
    "Healthcare",
    "Retail",
    "Manufacturing",
    "Telecommunications",
    "Education",
    "Energy",
    "Media",
    "Professional Services"
]


# ------------------------------------------------------------
# COMPANY NAME COMPONENTS
# ------------------------------------------------------------

COMPANY_PREFIXES = [
    "Apex", "Nova", "Vertex", "Global", "Pioneer",
    "Summit", "Bright", "Quantum", "NextGen", "Prime",
    "Blue", "Evergreen", "Unified", "Dynamic", "Future"
]

COMPANY_SUFFIXES = [
    "Technologies", "Solutions", "Systems", "Analytics",
    "Industries", "Consulting", "Innovations", "Digital",
    "Networks", "Enterprises", "Group", "Corporation"
]


# ------------------------------------------------------------
# COMPANY HEADQUARTERS
# ------------------------------------------------------------

LOCATION_COUNTRIES = {
    "New York": "USA",
    "San Francisco": "USA",
    "Seattle": "USA",
    "Austin": "USA",
    "Chicago": "USA",
    "London": "UK",
    "Toronto": "Canada",
    "Bengaluru": "India",
    "Mumbai": "India",
    "Delhi": "India",
    "Singapore": "Singapore",
    "Sydney": "Australia"
}

LOCATIONS = list(LOCATION_COUNTRIES.keys())


# ------------------------------------------------------------
# DEPARTMENTS
# ------------------------------------------------------------

DEPARTMENTS = [
    "Executive",
    "Engineering",
    "Data",
    "Product",
    "Sales",
    "Marketing",
    "Finance",
    "Operations",
    "Human Resources",
    "Legal",
    "Customer Success"
]


# ------------------------------------------------------------
# STRUCTURE-SPECIFIC WORKFORCE CONFIGURATION
# ------------------------------------------------------------

# These are structure probabilities, not manager assignments.
# Actual reporting relationships will be generated in a later step.

STRUCTURE_DESCRIPTION = {
    "Functional": "Employees organized around business functions",
    "Hierarchical": "Multiple management layers and reporting levels",
    "Flat": "Fewer management layers and wider reporting spans",
    "Divisional": "Independent business or product divisions",
    "Matrix-Inspired": "Cross-functional collaboration with one primary manager"
}


print("Company configuration loaded successfully.")
print("Company size categories:", len(COMPANY_SIZE_PROBABILITIES))
print("Industries:", len(INDUSTRIES))
print("Headquarters locations:", len(LOCATIONS))
print("Departments:", len(DEPARTMENTS))
print("Structure types:", len(STRUCTURE_DESCRIPTION))
```

```python

# ============================================================
# STEP 3: GENERATE SYNTHETIC COMPANIES
# ============================================================

# Company dataset schema
COMPANY_COLUMNS = [
    "company_id",
    "company_name",
    "industry",
    "company_size",
    "employee_count",
    "headquarters",
    "country",
    "organizational_structure"
]

companies = []

for i in range(NUM_COMPANIES):

    company_id = f"CMP{i+1:04d}"

    # Unique company name
    prefix = random.choice(COMPANY_PREFIXES)
    suffix = random.choice(COMPANY_SUFFIXES)
    company_name = f"{prefix} {suffix} {i+1:04d}"

    # Company size
    company_size = random.choices(
        list(COMPANY_SIZE_PROBABILITIES.keys()),
        weights=list(COMPANY_SIZE_PROBABILITIES.values()),
        k=1
    )[0]

    size_profile = COMPANY_SIZE_PROFILES[company_size]

    employee_count = random.randint(
        size_profile["min_employees"],
        size_profile["max_employees"]
    )

    # Company industry and headquarters
    industry = random.choice(INDUSTRIES)

    headquarters = random.choice(LOCATIONS)
    country = LOCATION_COUNTRIES[headquarters]

    # Organizational structure
    organizational_structure = random.choices(
        ORGANIZATIONAL_STRUCTURES,
        weights=STRUCTURE_WEIGHTS,
        k=1
    )[0]

    companies.append({
        "company_id": company_id,
        "company_name": company_name,
        "industry": industry,
        "company_size": company_size,
        "employee_count": employee_count,
        "headquarters": headquarters,
        "country": country,
        "organizational_structure": organizational_structure
    })


# Convert to DataFrame
companies_df = pd.DataFrame(
    companies,
    columns=COMPANY_COLUMNS
)


# ============================================================
# CORE PROJECT VALIDATION
# ============================================================

assert len(companies_df) == NUM_COMPANIES
assert companies_df["company_id"].is_unique
assert companies_df["company_name"].is_unique

assert (
    companies_df["employee_count"].min()
    >= MIN_EMPLOYEES_PER_COMPANY
)

assert (
    companies_df["employee_count"].max()
    <= MAX_EMPLOYEES_PER_COMPANY
)

assert (
    companies_df["employee_count"].sum()
    >= NUM_COMPANIES * MIN_EMPLOYEES_PER_COMPANY
)


# ============================================================
# GENERATION SUMMARY
# ============================================================

print("Company generation completed!")
print("Total companies:", len(companies_df))
print("Planned employees:", companies_df["employee_count"].sum())

print("\nCompany size distribution:")
print(companies_df["company_size"].value_counts())

print("\nOrganizational structure distribution:")
print(companies_df["organizational_structure"].value_counts())

print("\nSample companies:")
display(companies_df.head(10))
```

```python

# ============================================================
# STEP 5: EMPLOYEE GENERATION CONFIGURATION
# ============================================================

# ------------------------------------------------------------
# EMPLOYEE NAMES
# ------------------------------------------------------------

FIRST_NAMES = [
    "Aarav", "Vivaan", "Aditya", "Arjun", "Rohan",
    "Rahul", "Aman", "Karan", "Ishaan", "Kabir",
    "Priya", "Ananya", "Aisha", "Neha", "Riya",
    "Sneha", "Meera", "Diya", "Kavya", "Isha",
    "James", "John", "Michael", "David", "Daniel",
    "Emma", "Olivia", "Sophia", "Emily", "Grace",
    "Liam", "Noah", "Oliver", "Lucas", "Ethan",
    "Mia", "Charlotte", "Amelia", "Ella", "Chloe"
]

LAST_NAMES = [
    "Sharma", "Verma", "Gupta", "Singh", "Patel",
    "Mehta", "Reddy", "Kumar", "Joshi", "Malhotra",
    "Smith", "Johnson", "Brown", "Taylor", "Wilson",
    "Anderson", "Thomas", "Martin", "Clark", "Lewis",
    "Miller", "Davis", "White", "Harris", "Walker",
    "Thompson", "Moore", "Young", "Allen", "King"
]


# ------------------------------------------------------------
# BUSINESS DIVISIONS
# Used to create divisional organizational structures.
# ------------------------------------------------------------

DIVISION_NAMES = [
    "Consumer Products",
    "Enterprise Solutions",
    "Digital Services",
    "International Markets",
    "Business Operations"
]


# ------------------------------------------------------------
# TITLE-TO-DEPARTMENT MAPPING
# Each generated job title receives a primary department.
# ------------------------------------------------------------

def get_department(job_title):

    title = job_title.lower()

    if any(word in title for word in [
        "engineering", "engineer", "technical",
        "software", "technology"
    ]):
        return "Engineering"

    elif any(word in title for word in [
        "data", "analytics"
    ]):
        return "Data"

    elif any(word in title for word in [
        "product"
    ]):
        return "Product"

    elif any(word in title for word in [
        "sales"
    ]):
        return "Sales"

    elif any(word in title for word in [
        "marketing"
    ]):
        return "Marketing"

    elif any(word in title for word in [
        "finance", "financial"
    ]):
        return "Finance"

    elif any(word in title for word in [
        "operations"
    ]):
        return "Operations"

    elif any(word in title for word in [
        "human resources", "hr"
    ]):
        return "Human Resources"

    elif any(word in title for word in [
        "legal"
    ]):
        return "Legal"

    elif any(word in title for word in [
        "customer success"
    ]):
        return "Customer Success"

    elif any(word in title for word in [
        "chief executive"
    ]):
        return "Executive"

    else:
        return "General Management"


# ------------------------------------------------------------
# EMPLOYEE DATA CONTRACT
# No manager IDs or generation-only structure labels.
# ------------------------------------------------------------

EMPLOYEE_COLUMNS = [
    "employee_id",
    "company_id",
    "employee_name",
    "job_title",
    "department",
    "division"
]


# ------------------------------------------------------------
# CONFIGURATION VALIDATION
# ------------------------------------------------------------

assert len(FIRST_NAMES) > 0
assert len(LAST_NAMES) > 0

assert len(DIVISION_NAMES) >= 2

assert "employee_id" in EMPLOYEE_COLUMNS
assert "company_id" in EMPLOYEE_COLUMNS
assert "job_title" in EMPLOYEE_COLUMNS
assert "department" in EMPLOYEE_COLUMNS

assert "true_manager_id" not in EMPLOYEE_COLUMNS
assert "seniority_level" not in EMPLOYEE_COLUMNS

print("Employee-generation configuration loaded!")
print("First names:", len(FIRST_NAMES))
print("Last names:", len(LAST_NAMES))
print("Business divisions:", len(DIVISION_NAMES))
print("Employee input columns:", EMPLOYEE_COLUMNS)
```

```python

# ============================================================
# STEP 6: EMPLOYEE SENIORITY DISTRIBUTION
# ============================================================

# Convert workforce distribution into probabilities
WORKFORCE_CATEGORIES = list(WORKFORCE_DISTRIBUTION.keys())

WORKFORCE_PROBABILITIES = np.array(
    list(WORKFORCE_DISTRIBUTION.values()),
    dtype=float
)

# Normalize probabilities to ensure they sum to 1
WORKFORCE_PROBABILITIES = (
    WORKFORCE_PROBABILITIES /
    WORKFORCE_PROBABILITIES.sum()
)


def generate_employee_categories(employee_count):
    """
    Generate exactly one CEO and distribute all remaining
    employees across the other seniority categories.
    """

    if employee_count < 2:
        raise ValueError("A company must have at least two employees.")

    non_ceo_count = employee_count - 1

    category_counts = np.random.multinomial(
        non_ceo_count,
        WORKFORCE_PROBABILITIES
    )

    categories = ["CEO"]

    for category, count in zip(
        WORKFORCE_CATEGORIES,
        category_counts
    ):
        categories.extend([category] * int(count))

    # Shuffle non-CEO employees without moving the CEO
    remaining_categories = categories[1:]
    random.shuffle(remaining_categories)

    categories = ["CEO"] + remaining_categories

    assert len(categories) == employee_count
    assert categories.count("CEO") == 1

    return categories


# ============================================================
# TEST THE ALLOCATION FUNCTION
# ============================================================

sample_employee_count = 150

sample_categories = generate_employee_categories(
    sample_employee_count
)

sample_distribution = (
    pd.Series(sample_categories)
    .value_counts()
    .reindex(
        ["CEO"] + WORKFORCE_CATEGORIES,
        fill_value=0
    )
)

print("Sample company employee allocation:")
print(sample_distribution)

print("\nTotal employees:", len(sample_categories))
print("CEO count:", sample_categories.count("CEO"))

assert len(sample_categories) == sample_employee_count
assert sample_categories.count("CEO") == 1
```

```python

# ============================================================
# STEP 7: GENERATE EMPLOYEE POPULATION
# ============================================================

# ------------------------------------------------------------
# 1. DEPARTMENT MAPPING
# ------------------------------------------------------------

def get_department(job_title):

    title = job_title.lower()

    if "chief executive" in title:
        return "Executive"

    elif any(x in title for x in [
        "engineering", "engineer", "technical",
        "software", "technology"
    ]):
        return "Engineering"

    elif "data" in title or "analytics" in title:
        return "Data"

    elif "product" in title:
        return "Product"

    elif "sales" in title:
        return "Sales"

    elif "marketing" in title:
        return "Marketing"

    elif any(x in title for x in ["finance", "financial"]):
        return "Finance"

    elif "operat" in title:
        return "Operations"

    elif any(x in title for x in [
        "human resources", "hr"
    ]):
        return "Human Resources"

    elif "legal" in title:
        return "Legal"

    elif "customer success" in title:
        return "Customer Success"

    return "General Management"


# ------------------------------------------------------------
# 2. EMPLOYEE GENERATION
# ------------------------------------------------------------

employee_records = []

employee_counter = 1

for company in companies:

    company_id = company["company_id"]
    employee_count = company["employee_count"]

    structure = company["organizational_structure"]

    # Generate one CEO and allocate remaining categories
    employee_categories = generate_employee_categories(
        employee_count
    )

    for position, category in enumerate(employee_categories):

        # Select title from the assigned category
        job_title = random.choice(JOB_TITLES[category])

        # CEO and executives belong to corporate leadership
        if category in ["CEO", "Executive"]:
            division = "Corporate"

        elif structure == "Divisional":
            division = random.choice(DIVISION_NAMES)

        else:
            division = "Corporate"

        first_name = random.choice(FIRST_NAMES)
        last_name = random.choice(LAST_NAMES)

        employee_name = f"{first_name} {last_name}"

        employee_id = f"EMP{employee_counter:07d}"

        department = get_department(job_title)

        employee_records.append({

            "employee_id": employee_id,
            "company_id": company_id,
            "employee_name": employee_name,
            "job_title": job_title,
            "department": department,
            "division": division,

            # Generation metadata, not inference input
            "seniority_category": category,
            "seniority_level": SENIORITY_LEVELS[category],
            "organizational_structure": structure
        })

        employee_counter += 1


# ------------------------------------------------------------
# 3. CREATE GENERATION DATAFRAME
# ------------------------------------------------------------

employees_generation_df = pd.DataFrame(employee_records)


# ------------------------------------------------------------
# 4. CREATE INFERENCE INPUT DATAFRAME
# ------------------------------------------------------------

employees_input_df = employees_generation_df[
    EMPLOYEE_COLUMNS
].copy()


# ------------------------------------------------------------
# 5. HIERARCHY POPULATION SUMMARY
# ------------------------------------------------------------

print("Employee generation completed!")

print("Total employees:", len(employees_generation_df))

print(
    "Total companies:",
    employees_generation_df["company_id"].nunique()
)

print("\nSeniority distribution:")

display(
    employees_generation_df["seniority_category"]
    .value_counts()
    .reindex(SENIORITY_LEVELS.keys(), fill_value=0)
    .rename_axis("seniority_category")
    .reset_index(name="employee_count")
)

print("\nOrganizational structure distribution:")

display(
    employees_generation_df.groupby(
        "organizational_structure"
    ).size().reset_index(name="employee_count")
)


# ------------------------------------------------------------
# 6. CORE HIERARCHY VALIDATION
# ------------------------------------------------------------

ceo_summary = (
    employees_generation_df[
        employees_generation_df["seniority_category"] == "CEO"
    ]
    .groupby("company_id")
    .size()
)

assert len(employees_generation_df) == sum(
    company["employee_count"] for company in companies
)

assert employees_generation_df["employee_id"].is_unique

assert employees_generation_df["company_id"].nunique() == NUM_COMPANIES

assert len(ceo_summary) == NUM_COMPANIES

assert (ceo_summary == 1).all()

assert (
    employees_generation_df.loc[
        employees_generation_df["seniority_category"] == "CEO",
        "seniority_level"
    ] == 0
).all()

assert (
    employees_generation_df.loc[
        employees_generation_df["seniority_category"] != "CEO",
        "seniority_level"
    ] > 0
).all()

assert "true_manager_id" not in employees_input_df.columns

assert "seniority_level" not in employees_input_df.columns

assert "organizational_structure" not in employees_input_df.columns


print("\nAll employee hierarchy validations passed!")

print("Exactly one CEO per company.")
print("CEO seniority rank is zero.")
print("No manager IDs in inference input.")
print("Generation metadata is separated from inference input.")

print("\nSample inference records:")
display(employees_input_df.head(10))
```

```python
# ============================================================
# STEP 8B: STRUCTURE-AWARE GROUND-TRUTH GENERATION
# ============================================================

import random
import pandas as pd
from collections import defaultdict

hierarchy_rng = random.Random(RANDOM_SEED + 200)

generation_df = employees_generation_df.copy()

ground_truth_records = []

for company_id, company_df in generation_df.groupby("company_id", sort=False):

    structure = company_df["organizational_structure"].iloc[0]
    rules = STRUCTURE_RULES[structure]

    records = company_df.to_dict("records")

    employees_by_level = defaultdict(list)
    manager_pools = defaultdict(list)
    direct_report_counts = {}
    manager_capacity = {}

    for emp in records:
        employees_by_level[emp["seniority_level"]].append(emp)

        if emp["seniority_category"] in MANAGER_CATEGORIES:
            level = emp["seniority_level"]
            manager_pools[level].append(emp)

            manager_id = emp["employee_id"]
            direct_report_counts[manager_id] = 0
            manager_capacity[manager_id] = hierarchy_rng.randint(
                rules["min_reporting_span"],
                rules["max_reporting_span"]
            )

    ceos = [
        emp for emp in records
        if emp["seniority_category"] == "CEO"
    ]

    if len(ceos) != 1:
        raise ValueError(f"Expected exactly one CEO in {company_id}")

    ceo_id = ceos[0]["employee_id"]
    company_assignments = {ceo_id: None}

    for level in sorted(employees_by_level.keys()):

        if level == 0:
            continue

        level_employees = employees_by_level[level].copy()
        hierarchy_rng.shuffle(level_employees)

        for emp in level_employees:

            employee_id = emp["employee_id"]
            employee_department = emp["department"]
            employee_division = emp["division"]

            # Gather managers with higher seniority.
            eligible_managers = []

            for manager_level, pool in manager_pools.items():
                if manager_level < level:
                    eligible_managers.extend(pool)

            if not eligible_managers:
                raise ValueError(
                    f"No eligible manager for {employee_id} in {company_id}"
                )

            # Divisional companies: prefer managers from the same
            # division, while allowing Corporate-level oversight.
            if structure == "Divisional":
                divisional_candidates = [
                    manager for manager in eligible_managers
                    if (
                        manager["division"] == employee_division
                        or manager["division"] == "Corporate"
                    )
                ]

                if divisional_candidates:
                    eligible_managers = divisional_candidates

            manager_scores = []

            for manager in eligible_managers:

                manager_id = manager["employee_id"]
                seniority_gap = level - manager["seniority_level"]

                score = 1.0

                same_department = (
                    manager["department"] == employee_department
                )

                same_division = (
                    manager["division"] == employee_division
                )

                # Functional: strong preference for same department.
                if structure == "Functional":
                    score *= 8.0 if same_department else 0.5
                    score *= 1.0 / (1.0 + 0.15 * (seniority_gap - 1))

                # Hierarchical: favor the closest higher-seniority level.
                elif structure == "Hierarchical":
                    score *= 1.0 / (seniority_gap ** 2)
                    if same_department:
                        score *= 1.5

                # Flat: favor broader seniority gaps and skip-level links.
                elif structure == "Flat":
                    score *= float(seniority_gap ** 1.5)
                    if same_department:
                        score *= 1.2

                # Divisional: strongly favor same-division managers.
                elif structure == "Divisional":
                    if same_division:
                        score *= 12.0
                    elif manager["division"] == "Corporate":
                        score *= 3.0
                    else:
                        score *= 0.1

                    if same_department:
                        score *= 1.5

                # Matrix-Inspired: allow cross-functional primary reporting.
                elif structure == "Matrix-Inspired":
                    score *= 4.0 if not same_department else 1.0
                    score *= 1.0 / (1.0 + 0.10 * (seniority_gap - 1))

                # Soft workload preference.
                load_ratio = (
                    direct_report_counts[manager_id]
                    / manager_capacity[manager_id]
                )
                score *= 1.0 / (1.0 + load_ratio)

                manager_scores.append(max(score, 0.001))

            selected_manager = hierarchy_rng.choices(
                eligible_managers,
                weights=manager_scores,
                k=1
            )[0]

            selected_manager_id = selected_manager["employee_id"]

            company_assignments[employee_id] = selected_manager_id
            direct_report_counts[selected_manager_id] += 1

    for employee_id, manager_id in company_assignments.items():
        ground_truth_records.append({
            "employee_id": employee_id,
            "company_id": company_id,
            "true_manager_id": manager_id
        })

ground_truth_df = pd.DataFrame(
    ground_truth_records,
    columns=[
        "employee_id",
        "company_id",
        "true_manager_id"
    ]
)

print("Structure-aware ground-truth generation completed!")
print("Total records:", len(ground_truth_df))
print("Total companies:", ground_truth_df["company_id"].nunique())
print("Root CEOs:", ground_truth_df["true_manager_id"].isna().sum())
print(
    "Reporting relationships:",
    ground_truth_df["true_manager_id"].notna().sum()
)
```

```python
# ============================================================
# STEP 8C: VALIDATE REVISED GROUND-TRUTH HIERARCHY
# ============================================================

assert ground_truth_df["employee_id"].is_unique

assert set(ground_truth_df["employee_id"]) == set(
    employees_generation_df["employee_id"]
)

# Exactly one root CEO per company
root_counts = (
    ground_truth_df[
        ground_truth_df["true_manager_id"].isna()
    ]
    .groupby("company_id")
    .size()
)

assert len(root_counts) == 500
assert (root_counts == 1).all()

# Every non-root employee has a valid manager
non_root = ground_truth_df[
    ground_truth_df["true_manager_id"].notna()
]

manager_company = ground_truth_df[
    ["employee_id", "company_id"]
].rename(columns={
    "employee_id": "true_manager_id",
    "company_id": "manager_company_id"
})

manager_check = non_root.merge(
    manager_company,
    on="true_manager_id",
    how="left"
)

assert manager_check["manager_company_id"].notna().all()
assert (
    manager_check["company_id"]
    == manager_check["manager_company_id"]
).all()

# Every manager has higher seniority than their report
seniority_lookup = employees_generation_df[
    ["employee_id", "seniority_level"]
]

seniority_check = non_root.merge(
    seniority_lookup.rename(columns={
        "employee_id": "true_manager_id",
        "seniority_level": "manager_level"
    }),
    on="true_manager_id"
).merge(
    seniority_lookup.rename(columns={
        "seniority_level": "employee_level"
    }),
    on="employee_id"
)

assert (
    seniority_check["manager_level"]
    < seniority_check["employee_level"]
).all()

print("Revised hierarchy validation passed!")
print("Unique employee assignments:", ground_truth_df["employee_id"].nunique())
print("Companies with exactly one CEO:", len(root_counts))
print("Valid same-company relationships:", len(manager_check))
print("Correct seniority relationships:", len(seniority_check))
```

```python
# ============================================================
# STEP 8D: REVISED ORGANIZATIONAL STRUCTURE ANALYSIS
# ============================================================

# Employee-side attributes
employee_attributes = employees_generation_df[
    [
        "employee_id",
        "seniority_level",
        "department",
        "division",
        "organizational_structure"
    ]
]

# Manager-side attributes
manager_attributes = employees_generation_df[
    [
        "employee_id",
        "seniority_level",
        "department",
        "division"
    ]
].rename(columns={
    "employee_id": "true_manager_id",
    "seniority_level": "manager_level",
    "department": "manager_department",
    "division": "manager_division"
})

analysis_df = (
    ground_truth_df[
        ground_truth_df["true_manager_id"].notna()
    ]
    .merge(employee_attributes, on="employee_id")
    .merge(manager_attributes, on="true_manager_id")
)

analysis_df["seniority_gap"] = (
    analysis_df["seniority_level"]
    - analysis_df["manager_level"]
)

analysis_df["same_department"] = (
    analysis_df["department"]
    == analysis_df["manager_department"]
)

analysis_df["same_division"] = (
    analysis_df["division"]
    == analysis_df["manager_division"]
)

# Calculate direct reports per manager
manager_load = (
    analysis_df
    .groupby(
        ["organizational_structure", "true_manager_id"]
    )
    .size()
    .reset_index(name="direct_reports")
)

manager_load_summary = (
    manager_load
    .groupby("organizational_structure")
    .agg(
        avg_direct_reports=("direct_reports", "mean"),
        median_direct_reports=("direct_reports", "median")
    )
    .reset_index()
)

# Reporting-pattern summary
structure_summary = (
    analysis_df
    .groupby("organizational_structure")
    .agg(
        reporting_relationships=("employee_id", "count"),
        avg_seniority_gap=("seniority_gap", "mean"),
        same_department_pct=("same_department", "mean")
    )
    .reset_index()
)

structure_summary["same_department_pct"] *= 100

structure_summary = structure_summary.merge(
    manager_load_summary,
    on="organizational_structure"
)

print("Reporting patterns by organizational structure:")
display(structure_summary)
```

```python
# ============================================================
# STEP 8E: DIVISIONAL ALIGNMENT ANALYSIS
# ============================================================

divisional_df = analysis_df[
    analysis_df["organizational_structure"] == "Divisional"
].copy()

divisional_df["division_alignment"] = (
    (divisional_df["division"] == divisional_df["manager_division"])
    |
    (divisional_df["manager_division"] == "Corporate")
)

divisional_summary = (
    divisional_df
    .groupby("manager_division")
    .agg(
        reporting_relationships=("employee_id", "count"),
        exact_division_matches=("same_division", "sum"),
        aligned_relationships=("division_alignment", "sum")
    )
    .reset_index()
)

divisional_summary["division_alignment_pct"] = (
    divisional_summary["aligned_relationships"]
    / divisional_summary["reporting_relationships"]
    * 100
)

print("Divisional reporting alignment:")
display(divisional_summary)

print(
    "Overall division alignment:",
    round(divisional_df["division_alignment"].mean() * 100, 2),
    "%"
)
```
